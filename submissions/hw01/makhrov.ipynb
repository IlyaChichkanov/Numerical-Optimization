## Задача 2. Практика (6 баллов)

**(а) Линейный МНК и переобучение (3 балла).** Ниже сгенерированы обучающая и тестовая выборки: $y = \sin(2\pi t) + \varepsilon$, $\varepsilon \sim \mathcal{N}(0, 0.1^2)$. Для каждой степени $d = 1, \dots, 15$ подгоните полином $\varphi(t; x) = x_0 + x_1 t + \dots + x_d t^d$ методом наименьших квадратов (`numpy.linalg.lstsq`; матрицу $A$ удобно строить через `np.vander(t, d + 1, increasing=True)`). Постройте график зависимости среднеквадратичной ошибки на обучающей и на тестовой выборках от $d$ (ось $y$ — логарифмическая). Объясните поведение обеих кривых в двух-трёх предложениях: почему одна монотонна, а другая нет?

```python
N = 50
t_train = np.sort(rng.uniform(0, 1, N))
y_train = np.sin(2 * np.pi * t_train) + 0.1 * rng.standard_normal(N)
t_test = np.sort(rng.uniform(0, 1, N))
y_test = np.sin(2 * np.pi * t_test) + 0.1 * rng.standard_normal(N)

plt.figure(figsize=(5, 3))
plt.plot(t_train, y_train, "o", ms=4, label="обучающая")
plt.plot(t_test, y_test, "s", ms=4, label="тестовая")
plt.legend(); plt.xlabel("$t$"); plt.ylabel("$y$"); plt.title("Данные задачи 2(а)")
plt.show()
```

```python
degrees = range(1, 16)
train_mse, test_mse = [], []

for d in degrees:
    A_train = np.vander(t_train, d + 1, increasing=True)
    x, *_ = np.linalg.lstsq(A_train, y_train, rcond=None) 

    y_train_pred = A_train @ x
    A_test = np.vander(t_test, d + 1, increasing=True)
    y_test_pred = A_test @ x

    train_mse.append(np.mean((y_train_pred - y_train) ** 2))
    test_mse.append(np.mean((y_test_pred - y_test) ** 2))
```

**Ответ:** Обучающая ошибка монотонно убывает с ростом $d$, так как пространство полиномов степени $\le d$ вложено в пространство степени $\le d + 1$, и МНК не может дать худшее приближение на тех же точках при добавлении степеней свободы. Тестовая ошибка сначала падает вместе с обучающей (снижается смещение — модель лучше улавливает форму \sin(2\pi t)), но при больших $d$ лишние степени свободы подстраиваются под шум конкретной обучающей выборки (растёт дисперсия), поэтому тестовая ошибка перестаёт монотонно убывать и начинает расти/колебаться — переобучение.

**(б) LP: диета (3 балла).** Решите задачу диеты из 1(б) с помощью `scipy.optimize.linprog` для данных ниже (4 продукта, 3 вещества). Выведите оптимальное $x$ и стоимость. Определите, какие из ограничений $Ax \ge b$ и $x \ge 0$ **активны** в решении (невязка равна нулю с точностью $10^{-9}$). Затем увеличьте требование по первому веществу на единицу ($b_1 \to b_1 + 1$), решите заново и сравните: как изменились решение и стоимость? Изменился ли набор активных ограничений?

Подсказка: `linprog` минимизирует $c^\top x$ при $A_{ub} x \le b_{ub}$, поэтому ограничение $Ax \ge b$ нужно переписать со знаком минус; `bounds=(0, None)` задаёт $x \ge 0$.

```python
c = np.array([2.0, 3.0, 1.5, 4.0])                # цены продуктов
A = np.array([[1.0, 2.0, 0.0, 3.0],
              [2.0, 1.0, 1.0, 1.0],
              [0.0, 3.0, 2.0, 2.0]])                 # содержание вещества i в продукте j
b = np.array([10.0, 8.0, 12.0])                     # минимальные требования
```

```python
def solve_diet(b_vec, label):
    res = linprog(c, A_ub=-A, b_ub=-b_vec, bounds=(0, None), method="highs")
    x, cost = res.x, res.fun

    slack_A = A @ x - b_vec          
    active_A = np.isclose(slack_A, 0, atol=1e-9)
    active_x = np.isclose(x, 0, atol=1e-9)

    print(f"--- {label} ---")
    print("x* =", x)
    print("стоимость =", cost)
    print("активные Ax>=b (по веществам):", active_A)
    print("активные x>=0 (по продуктам): ", active_x)
    return x, cost, active_A, active_x

x1, cost1, aA1, ax1 = solve_diet(b, "b исходное")

b2 = b.copy()
b2[0] += 1
x2, cost2, aA2, ax2 = solve_diet(b2, "b1 -> b1+1")

print("Δx =", x2 - x1)
print("Δcost =", cost2 - cost1)
```

**Ответ:** 
Решение сместилось: закупки продуктов 1 и 2 немного уменьшились, а продукт 4 стал невыгодно игнорировать и он появился в корзине (0.667 ед.); стоимость выросла на $\Delta =1.111$ - это и есть по сути теневая цена (множитель Лагранжа) ограничения по первому веществу на новом решении. Набор активных ограничений по веществам не изменился (все три остались активными "в притык"), но набор активных ограничений неотрицательности $x\ge 0$ изменился — базис оптимального решения поменялся, что обычно для LP при возмущении правой части: решение "скользит" по новой грани допустимого многогранника, пока не появится новая активная переменная.
